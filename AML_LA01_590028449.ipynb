## B1. Scaling changes the answer  &nbsp;&nbsp;`[4 marks]`

*Dataset: A1 (California Housing).*

a. Split 80/20 with `random_state=0`. Train a `KNeighborsRegressor(n_neighbors=5)` on the raw features and report test MAE.
b. Repeat inside a `Pipeline` with `StandardScaler`. Report test MAE again.
c. Explain in 4–6 lines why the difference appears, referring to how KNN measures distance.


```python
# B1 — code
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error
# Load A1 dataset
X, y = fetch_california_housing(return_X_y=True, as_frame=True)
# 80/20 split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
# a. KNN on raw features
knn_raw = KNeighborsRegressor(n_neighbors=5)
knn_raw.fit(X_train, y_train)
y_pred_raw = knn_raw.predict(X_test)
mae_raw = mean_absolute_error(y_test, y_pred_raw)
print("Test MAE (Raw Features):", mae_raw)
# b. KNN with StandardScaler
knn_scaled = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsRegressor(n_neighbors=5))
])
knn_scaled.fit(X_train, y_train)
y_pred_scaled = knn_scaled.predict(X_test)
mae_scaled = mean_absolute_error(y_test, y_pred_scaled)
print("Test MAE (Standardized):", mae_scaled)
```

**Observation (B1):**

> *KNN measures the distance between data points using their feature values, typically Euclidean distance.
When features have different scales, features with larger numerical values can dominate the distance calculation.
In California Housing, Population has values much larger than features such as AveRooms and AveOccup.
StandardScaler puts the features on a comparable scale by standardizing them.
Therefore, the standardized KNN usually gives a better MAE because distance is based more fairly on all features.
